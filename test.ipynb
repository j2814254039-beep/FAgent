```python
import tkinter
from itertools import cycle
from pathlib import Path

from app.config import getActionPath

ImagePath = Path(getActionPath())
root = tkinter.Tk()
root.geometry("600x800")     # ← 和画布尺寸一致

FRAME_INTERVAL_MS = 100
_photo = None
# 背景透明
KEY = "#3c5a8c"
root.wm_attributes("-transparentcolor", KEY) # 置顶
root.configure(bg=KEY)

root.overrideredirect(True)# 去边框
label = tkinter.Label(root, bg=KEY, bd=0) # 去标签的边框
label.pack()
_drag = {}
def onPress(e):
    _drag["x"] = e.x_root - root.winfo_x()
    _drag["y"] = e.y_root - root.winfo_y()

def onDrag(e):
    root.geometry(f"+{e.x_root - _drag['x']}+{e.y_root - _drag['y']}")

def showMenu(e):
    menu = tkinter.Menu(root,tearoff=0)
    menu.add_command(label="退出",command=root.destroy)



label.bind("<Button-1>", onPress)
label.bind("<B1-Motion>", onDrag)
label.bind("<Button-3>", showMenu)

def walk():
    global _photo
    walkPath = ImagePath / "walk"
    frames = sorted(walkPath.glob("*.png"))
    if not frames:
        raise FileNotFoundError("走路图片库不存在")
    frameCycle = cycle(frames)
    def tick():
        global _photo
        frame = next(frameCycle)
        _photo = tkinter.PhotoImage(file=str(frame))
        label.configure(image=_photo)
        root.after(FRAME_INTERVAL_MS, tick)
    tick()
walk()
root.mainloop()
```
